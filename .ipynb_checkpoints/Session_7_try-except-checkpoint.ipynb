**try-except**

- try except is used to handle the exeption mechanism

- try except is used to catch and handle the errors

- when we get the error python will stop the execution 

- using try except eventhough we get the errors, the execution will not stop

- it will catch all the logical errors automatically

- the actual code block will write in **try block**

- if any error occures in try block, will catch the errors in **except** block 

- Syntax errors will not catch

```python
# selct === tab
try:
    salary=eval(input('enter the salary:'))
    tax_per=eval(input('enter the tax_per:'))
    tax= salary*tax_per/100
    print(f'the total tax is:{tax}')
except Exception as e:
    print(e)
```

```python
# even - odd code
try:
    num=eval(input('enter the num:'))
    if num%2==0:
        print(f'{num} is an even')
    else:
        print(f'{num} is an odd')
except Exception as e:
    print(e)
```

**try-except-finally**

- original code in **try** block

- error catch in **except** block

- **finally** block will actiavte either or try and except activate

- assume that try block exceuted then finally block can also exceute

- assume that except block exceuted then also finally block exceute

coffee well === try
clean the kitchen == finally

cofee gone === except
clean the kitchen === finally



```python
try:
    print('connection established')
except Exception as e:
    print(e)
finally:
    print('Connection closed')
```

```python
try:
    100/0
except Exception as e:
    print(e)
    print('error gaya isliye my agaya')
finally:
    print('Connection closed')
```
