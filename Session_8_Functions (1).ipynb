**args-kwargs**

- args means we can provide multiple value with out argument name

    - which means while calling function we can provide directly 'virat',102

    - while define the function you no need to mention argument names also

    - the output will comes in **tuple** format : ('virat',102)

- kwargs keyword argumnets means we can provide multiple values with argumen name

    - while define the function no need to provide the name of the argumnets

    - while call the functions definitely we need to provide with argument names

        - name='virat',runs=102

    - the output comes in dict format: {'name':'virat','runs':102}

```python
# def fun1(a,b):
#     print(a)
#     print(b)
# fun1(10,20)

def fun1(*args):
    print(args)
fun1(10,20) 
# with out name 
# output in tuple format
```
